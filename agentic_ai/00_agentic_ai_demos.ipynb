### What is a Token in LLM ,How to check the token count using Python, how to encode & decode tokens


- **Tokens** are **words, subwords, and characters** converted into a format that is easier for a language model to understand.
- A token can be:
  - A single character  
  - Part of a word (e.g., `-ing`, `-ed`)  
  - A full word etc.
- Each model has a **maximum token capacity**, called the **context window**,  
which defines the total number of tokens it can handle across **input and output combined**.
---
## Rule of Thumb – Tokens vs Characters

- On average:
  - **1 token ≈ 4 characters (English text)**
- This is an **approximation**, not an exact rule.
---
##### Example 1: Gemini 3 Pro
- **Context window:** ~1,000,000 tokens  
- **Max output tokens:** ~64,000 tokens  

**What this means:**
- You can send large inputs (books, repositories, transcripts)

##### Example 2:   GPT-5.2
- **Context window:** 400,000 tokens  
- **Max output tokens:** 128,000  

**What this means:**
- Total tokens (input + output) ≤ **400k**
- If you generate the maximum output:
  - Input tokens allowed ≈ **272k**
  - Output tokens = **128k**

## tiktoken

- **tiktoken** is a fast, open-source **tokenizer** developed by **OpenAI**.
- It is used to convert raw text into **tokens** that language models can process.
- A tokenizer works based on a specific **encoding** (e.g., `cl100k_base`).
- The same text can be tokenized **differently** depending on the encoding used.

# Pydantic

Pydantic is a widely used data validation library for Python.

Pydantic allows you to define structured, validated, and reliable data models using simple Python syntax — making it essential for modern APIs and Agentic AI workflows.
```bash
uv add pydantic
```

## Why Use It?

- Reduces boilerplate code (no need to write manual checks like `if type(x) != int...`)
- Excellent integration with FastAPI and AI frameworks (OpenAI, LangChain)
- High performance (Rust-based core in V2)

## BaseModel Class

**BaseModel** is the parent class that you inherit from to create your own data models.

**Role:** It handles all the logic for parsing, validating, and converting the data you feed into your class.
```python
from pydantic import BaseModel, Field
```

## Field

**Field** is a function used to customize validation and metadata for a specific variable within your model.

**Role:** It allows you to set constraints (e.g., `gt=0` for "greater than 0"), add descriptions (useful for documentation), or handle default values.

## Literal

**Literal** is used when a field must be one of a specific set of fixed values.

## Advantages of Pydantic

- **Data Validation:** Automatically ensures data is correct before your code processes it
- **Type Conversion:** Intelligently converts types (e.g., input string `"22"` becomes integer `22`)
- **Editor Support:** Uses standard Python type hints, providing excellent autocomplete in IDEs like VS Code
- **JSON Serialization:** Easily converts objects to JSON (useful for APIs)
- **Performance:** Core validation logic is written in Rust, making it extremely fast
